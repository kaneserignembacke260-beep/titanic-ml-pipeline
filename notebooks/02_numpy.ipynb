```python
def calculer_moyenne(liste):
    return sum(liste) / len(liste)
ages = [29, 41, 22, 35]
print(calculer_moyenne(ages))
```

### Calcul manuel de la moyenne

Cette première partie démontre comment calculer la moyenne d'une liste de nombres en utilisant les fonctions Python natives `sum()` et `len()`.

```python
import numpy as np

ages = np.array([29, 41, 22, 35])
print(np.mean(ages))
```

### Calcul de la moyenne avec NumPy

Cette seconde partie utilise la bibliothèque `NumPy`, qui est largement recommandée pour les opérations numériques en Python, notamment pour son efficacité et sa concision. La fonction `np.mean()` est optimisée pour ce type de calcul, surtout avec des tableaux de grandes dimensions.
