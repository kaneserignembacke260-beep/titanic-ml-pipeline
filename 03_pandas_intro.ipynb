```python
# describe() : donne un résumé statistique des colonnes numériques du dataset
# - count : nombre de valeurs non manquantes
# - mean  : moyenne (somme des valeurs / nombre de valeurs)
# - std   : écart-type = dispersion des valeurs autour de la moyenne
# - min / max : valeur la plus petite / la plus grande
# - 25% / 50% / 75% : quartiles → valeurs qui divisent le dataset trié
#   (50% = médiane, la valeur du milieu)
df.describe()
```

```python
df["Sex"].value_counts()
```

```python
df.groupby("Sex")["Survived"].mean()
# df.groupy("col") ici sexe permet de regrouper toutes  les lignes en  2  homme et femme
# on regarde que la colonne des survived
#  on calcule la moyenne à partir de male et femelle
```

```python
df.groupby("Pclass")["Survived"].mean()
# calcul le pourcentage dess  classes (1,2,3)
#  on constate que la classe 1 a la plus grande pourcentage de survie
```
